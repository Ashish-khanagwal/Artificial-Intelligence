## Generator

```python
def square(n):
  for i in range(3):
    yield i**2
  
square(3)
```

```python
for i in square(3):
  print(i)
```

```python
a = square(3)
next(a)
```

```python
def read_large_file(filepath):
  with open(filepath, 'r') as file:
    line = file.read()
    for file in line:
      yield line

file_path = 'data.txt'
for line in read_large_file(file_path):
  print(line.strip())
```
