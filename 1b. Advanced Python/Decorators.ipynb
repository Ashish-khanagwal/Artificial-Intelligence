```python
# Function Copy

def Welcome():
  return 'Welcome'

Welcome()
```

```python
wel=Welcome
print(wel())
del Welcome
print(wel())
```

```python
# Closures
def outer():
  msg = 'Hello'
  def inner():
    print("Hiiii")
    print(msg)
    print("How are you?")

  return inner()

outer()
```

```python
def outer(func):
  def inner():
    print('hii')
    func('heyyyy')
    print('how are you?')

  return inner()

outer(print)
```

```python
def outer(func, lst):
  def inner():
    print('length of list is:')
    print(func(lst))

  return inner()
outer(len, [1,2,3,4,5])
```

```python
def main_func(func):
  def child_func():
    print("hiii")
    func()
    print("See above message")
  return child_func()
```

```python
def chain_func():
  print("Yes, see me!!")

main_func(chain_func)
```

```python
## Decorator

@main_func
def chain_func():
  print("Yes, see me!!")
```
