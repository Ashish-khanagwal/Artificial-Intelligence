### Assignment 11: Abstract Base Class

Create an abstract base class named `Shape` with an abstract method `area`. Create derived classes `Circle` and `Square` that implement the `area` method. Create objects of the derived classes and call the `area` method.

```python
from abc import ABC, abstractmethod

class Shape:
  
  @abstractmethod
  def area(self):
    pass

class Circle(Shape):
  def area(self, radius):
    return 3.14*radius*radius
  
class Square(Shape):
  def area(self, side):
    return side*side
  
c1 = Circle()
print(c1.area(5))

s1 = Square()
print(s1.area(5))
```

### Assignment 12: Operator Overloading

Create a class named `Vector` with attributes `x` and `y`. Overload the `+` operator to add two `Vector` objects. Create objects of the class and test the operator overloading.

```python
class Vector:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __add__(self, other):
    return Vector(self.x+other.x, self.y+other.y)
  
  def __repr__(self):
    return f"Vetcor: {self.x}, {self.y}"
  
v1 = Vector(2, 4)
v2 = Vector(5, 6)

print(v1+v2)
```

### Assignment 13: Class with Custom Exception

Create a custom exception named `InsufficientBalanceError`. In the `BankAccount` class, raise this exception when a withdrawal amount is greater than the balance. Handle the exception and print an appropriate message.


```python
class Error(Exception):
  pass

class InsufficientBalanceError(Error):
  pass

class BankAccount:
  def __init__(self, owner, balance):
    self.owner = owner
    self.__balance = balance

  def get_balance(self):
    return f"Your balance is: {self.__balance}"
  
  def deposit(self, amount):
    self.__balance += amount
  
  def withdraw(self, amount):
    if self.__balance < amount:
      raise InsufficientBalanceError('Insufficient funds in account!')
    else:
      self.__balance -= amount

customer1 = BankAccount('Ashish', 1000000)
print(customer1.get_balance())
customer1.deposit(50000)
print(customer1.get_balance())
try:
    customer1.withdraw(1900000)
except InsufficientBalanceError as e:
    print(f"Error: {e}")
```

### Assignment 14: Class with Context Manager

Create a class named `FileManager` that implements the context manager protocol to open and close a file. Use this class to read the contents of a file.

```python
class FileManager:
  def __init__(self, filename, mode):
    self.filename = filename
    self.mode = mode

  def __enter__(self):
    self.file = open(self.filename, self.mode)
    return self.file
  
  def __exit__(self, exc_type, exc_value, traceback):
    self.file.close()
  
with FileManager('sample.txt', 'r') as file:
  content = file.read()
  print(content)
```

### Assignment 15: Chaining Methods

Create a class named `Calculator` with methods to add, subtract, multiply, and divide. Each method should return the object itself to allow method chaining. Create an object and chain multiple method calls.

```python
class Calculator:
  def __init__(self, value=0):
    self.value = value

  def add(self, amount):
    self.value += amount
    return self

  def subtract(self, amount):
    self.value -= amount
    return self

  def multiply(self, amount):
    self.value *= amount
    return self

  def divide(self, amount):
    if amount != 0:
      self.value /= amount
    else:
      print("Cannot divide by zero!")
    return self

calc = Calculator()
calc.add(10).subtract(3).multiply(2).divide(2)
print(calc.value) 
```
