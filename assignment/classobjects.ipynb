### Assignment 1: Basic Class and Object Creation

Create a class named `Car` with attributes `make`, `model`, and `year`. Create an object of the class and print its attributes.

```python
class Car:
  def __init__(self, make, model, year):
    self.make = make
    self.model = model
    self.year = year

car1 = Car('Audy', 'Q6', '2027')
print(car1.make)
print(car1.model)
print(car1.year)
```

### Assignment 4: Class with Private Attributes

Create a class named `BankAccount` with private attributes `account_number` and `balance`. Add methods to deposit and withdraw money, and to check the balance. Create an object of the class and perform some operations.

```python
class BankAccount:
  def __init__(self, owner, account_number, balance=0):
    self.owner = owner
    self.__account_number = account_number
    self.balance = balance
  
  def deposit(self, amount):
    self.balance += amount
    print(f"{amount} is deposited in account with account number {self.__account_number}. New balance is {self.balance}")

  def get_accountnumber(self):
    return self.__account_number

  def withdraw(self, amount):
    if amount > self.balance:
      print(f"Insuficient balance in account with account number {self.__account_number}")
    else:
      self.balance -= amount
      print(f"{amount} is withdrawned from account with account number {self.__account_number}. New balance is {self.balance}")

  def get_amount(self):
    return self.balance
  
account1 = BankAccount('Ashish', 123456, 1000000)
print(account1.owner)
print(account1.get_accountnumber())
print(account1.balance)

account1.get_accountnumber()
account1.deposit(50000)

account1.withdraw(40000)
```
