````python
def fib(n, memo={}):
  if n in memo:
    return memo[n]
  
  if n <= 1:
    return n
  memo[n] = fib(n-1, memo) + fib(n-2, memo)
  return memo[n]

print(fib(10))
print(fib(15))
````

### Assignment 6: Function with Decorators

Define a function that calculates the time taken to execute another function. Apply this decorator to a function that performs a complex calculation. Test the decorated function with different inputs.

# 🐍 What Is a Decorator? (Python Documentation — Simplified)

A decorator in Python is:

A function that takes another function, adds some extra ability to it, and returns a new improved function — without changing the original function’s code.

Think of it as putting a wrapper around a gift:
The gift is still inside, but now it looks nicer, behaves differently, or has extra features.

# 🎁 Analogy: Decorator = Gift Wrapper

Suppose you buy a simple gift (your function).
You take it to a gift-wrapping shop.

The shop:

- takes your gift (callback function)
- wraps it (adds extra behavior)
- returns it back to you as a wrapped version

Now, when someone opens the gift:

- they still get the same thing
- but with extra “before” and “after” effects (ribbon, bow, message tag, etc.)
- That’s exactly how decorators work.

````python
import time

def timer_decorator(func):
  def wrapper_f(*args, **kwargs):
    start_time = time.time()
    result = func(*args, **kwargs)
    end_time = time.time()
    print(f"{func.__name__} tool {end_time - start_time:.5f} seconds to execute")
    return result
  return wrapper_f

@timer_decorator
def complex_function(n):
  return sum(x**2 for x in range(n))

print(complex_function(10000))
````

### Assignment 7: Higher-Order Function for Filtering and Mapping

Define a higher-order function that takes two functions, a filter function and a map function, along with a list of integers. The higher-order function should first filter the integers using the filter function and then apply the map function to the filtered integers. Test with different filter and map functions.

# 🐍 Higher-Order Function for Filtering and Mapping

A higher-order function (HOF) is simply:

A function that takes another function as input, or returns a function.

In this problem, you want a function that:
- filters items using one callback
- maps/transforms items using another callback
- returns the final processed list

# 🧩 Analogy: The Fruit Factory

Imagine you have a fruit factory:

- The filter callback decides which fruits to keep
(e.g., “keep only ripe fruits”)

- The map callback decides how to transform them
(e.g., “cut fruits into slices”)

The higher-order function is the factory machine that:

- accepts fruits (list)
- accepts a selection rule (filter function)
- accepts a transformation rule (map function)
- produces processed fruits (output list)

````python
def filter_and_map(filter_func, map_func, lst):
  return [map_func(x) for x in lst if filter_func(x)]

print(filter_and_map(lambda x: x % 2 == 0, lambda x: x*2, [1, 2, 3, 4, 5, 6, 7, 8]))
````

# 🐍 What is functools.partial?

`partial()` lets you pre-fill some arguments of a function, creating a new function with fewer arguments.

Think of it as:

Saving a version of a function with some parameters already decided.

# 🎯 Analogy: Coffee Machine

Imagine a coffee machine function:

```
make_coffee(size, sugar, milk)
```


If you always drink medium, 1-sugar coffee, you can create a preset:

```
medium_sweet = partial(make_coffee, "medium", 1)
```


Now calling:

```
medium_sweet(milk=True)
```


works as:

```
make_coffee("medium", 1, milk=True)
```


You've locked the first two arguments.

```
from functools import partial
```

````python
from functools import partial

def multiply(n, m):
  return n*m

multiple_2 = partial(multiply, 2)
multiple_3 = partial(multiply, 3)
print(multiple_2(4))
print(multiple_3(4))
````

### Assignment 11: Function with Generators

Define a function that generates an infinite sequence of Fibonacci numbers. Test by printing the first 10 numbers in the sequence.

# 🐍 What is a Generator Function?

A generator function is a special kind of function that does not return all values at once.
Instead, it yields values one at a time, only when needed.

Think of it as:

A lazy sequence producer.

Rather than cooking all food at once, it cooks each item only when someone asks.

# 🍔 Analogy: The Burger Stand

Imagine a burger stand:

- Customers order burgers one by one
- The chef cooks each burger only when asked
- Chef says “Here’s your burger!” → yield
- The process pauses
- Next customer arrives → cooking continues
- That’s exactly what a generator does.

# 📌 Defining a Generator Function

A function becomes a generator when it uses:

`yield`


instead of:

`return`


Example:
```python
def counter():
    n = 1
    while n <= 5:
        yield n
        n += 1
```

Calling:

`c = counter()`


Does not run the function — it returns a generator object.

Then:

`next(c)`


Gives values one-by-one.

🚀 Example 1: Simple generator
```python
def squares(n):
    for i in range(1, n + 1):
        yield i * i

gen = squares(5)

for value in gen:
    print(value)
```

```
Output:
1
4
9
16
25
```

````python
def fibonacci_generator():
  a, b = 0, 1
  while True:
    yield a
    a, b = b, a+b

fib_gen = fibonacci_generator()
for _ in range(10):
  print(next(fib_gen), end=" ")
````

### Assignment 12: Currying

Define a curried function that takes three arguments, one at a time, and returns their product. Test the function by providing arguments one at a time.

# 🐍 What is Currying? (Simple Definition)

Currying means:

Turning a function that takes multiple arguments
into a chain of functions,
each taking one argument at a time.

So instead of:

`f(a, b, c)`


you transform it into:

`f(a)(b)(c)`


Each function returns another function until all arguments are provided.

# 🍕 Analogy: Build Your Pizza Step-by-Step

Imagine ordering pizza:

Normal function (non-curried):

You give all details at once:

"Large, Thin Crust, Extra Cheese"

So:

`order(size, crust, topping)`

#### Curried version:

You choose one thing per step:
```
order(size) → gives a function waiting for crust  
order(size)(crust) → gives a function waiting for topping  
order(size)(crust)(topping) → final order confirmed
```

Currying = ordering step-by-step.

### 🧩 Currying in Python (Manual Version)

Python doesn’t do currying by default like Haskell,
but you can build it:
```python
def curried_add(a):
    def inner1(b):
        def inner2(c):
            return a + b + c
        return inner2
    return inner1

print(curried_add(1)(2)(3))
```

Output:

`6`


Each call returns the next function until all arguments are consumed.

### ✔️ Cleaner Version (Two Arguments)
```python
def multiply(a):
    def inner(b):
        return a * b
    return inner

double = multiply(2)
print(double(5))
```

Output:
`10`


This is a curried function:

First call sets a = 2

Second call gives the result

## 🔥 Currying With `functools.partial`

`partial()` helps create curried-like behavior.

Example:
```python
from functools import partial

def power(base, exp):
    return base ** exp

square = partial(power, exp=2)
cube   = partial(power, exp=3)

print(square(5))  # 25
print(cube(3))    # 27
```

This is partial currying because one argument is fixed first.

## 🧠 Currying vs Partial Application (Very Important)
| Concept             | What it means            | Example         |
| ------------------- | ------------------------ | --------------- |
| Currying            | Convert f(a,b) → f(a)(b) | f(2)(3)         |
| Partial Application | Pre-fill only some args  | partial(f, a=2) |


````python
def curried_func(a):
  def inner_1(b):
    def inner_2(c):
      return a + b + c
    return inner_2
  return inner_1

print(curried_func(2)(3)(4))
````

### Assignment 13: Function with Context Manager

Define a function that uses a context manager to write a list of integers to a file. The function should handle any errors that occur during file operations. Test with different lists.

````python
def write_to_file(lst, filename):
  try:
    with open(filename, 'w') as f:
      for num in lst:
        f.write(f"{num}\n")
  except IOError as e:
    print(f"An error occurred: {e}")

# Test
write_to_file([1, 2, 3, 4, 5], 'output.txt')
````

### Assignment 14: Function with Multiple Return Types

Define a function that takes a list of mixed data types (integers, strings, and floats) and returns three lists: one containing all the integers, one containing all the strings, and one containing all the floats. Test with different inputs.

````python
def separate_types(lst):
    ints, strs, floats = [], [], []
    for item in lst:
        if isinstance(item, int):
            ints.append(item)
        elif isinstance(item, str):
            strs.append(item)
        elif isinstance(item, float):
            floats.append(item)
    return ints, strs, floats

# Test
print(separate_types([1, 'a', 2.5, 3, 'b', 4.0, 'c']))  # ([1, 3], ['a', 'b', 'c'], [2.5, 4.0])
````
