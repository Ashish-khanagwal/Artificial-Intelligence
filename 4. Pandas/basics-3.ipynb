## Converting the datatype of a particular column.

```python
import pandas as pd
import numpy as np
```

```python
# Creating DataFrames using a dictionary
people = {
    "first": [
        "Ashish", "Corey", "John", "Meera", "Arjun",
        "Sophia", "Liam", "Hannah", "Carlos", "Jisoo"
    ],
    "last": [
        "Khanagwal", "Anderson", "Doe", "Patel", "Singh",
        "Williams", "Brown", "Nguyen", "Garcia", "Kim"
    ],
    "email_address": [
        "ashishkhanagwal@gmail.com",
        "coreyschafar@gmail.com",
        "timdavid@gmail.com",
        "meerapatel@gmail.com",
        "arjunsingh@gmail.com",
        "sophiawilliams@gmail.com",
        "liambrown@gmail.com",
        "hannahnguyen@gmail.com",
        "carlosgarcia@gmail.com",
        "jisookim@gmail.com"
    ],
    "age": [24, 27, 26, 25, 29, 31, 28, 24, 32, 26],
    "Profession": [
        "AI Engineer", "Tutor", "Software Developer", "Data Scientist", "Product Manager",
        "UX Designer", "DevOps Engineer", "Research Analyst", "Systems Architect", "ML Engineer"
    ],
    "Company": [
        "Meta", "YouTube", "Google", "Microsoft", "Amazon",
        "Apple", "Netflix", "Tesla", "IBM", "OpenAI"
    ],
    "Score": [85, 90, 88, 92, 89, 87, 91, 86, 93, 95]
}
```

```python
df = pd.DataFrame(people)
df
```

### 1. Using .astype()
- This is the most common and easiest way.

```python
df['age'] = df['age'].astype(str)
# This will covert data type of age from int to string.
df['age']

df['Score'] = df['Score'].astype(float)
# This will convert the data type of score from int to float.
df['Score']
```

### 2. Using `pd.to_numeric()`, `pd.to_datetime()`, `pd.to_timedelta()`
- These are special functions for certain datatypes.

```python
# To numbers (int/float):
df['age'] = pd.to_numeric(df['age'])
df['age']
# Converted from string to integer.
```

```python
# To check data types of the dataframe.
df.dtypes
```

## Adding and Removing rows and columns
Adding and removing rows and columns in a pandas DataFrame is a common operation when preparing or cleaning up your data. Here’s how to do it in a clear, friendly way with practical examples:

## Adding Columns
Assign a new column by name:

```python
df['full name'] = df['first'] + ' ' + df['last']
```

```python
df
```

## Removing Columns
Using drop:

```python
df.drop(columns=['first', 'last'], inplace = True)
df
# This will remove the 'first' and 'last' columns from the dataframe.
# Removes the column “column_name” from the DataFrame. Use axis=1 for columns.
```

```python
df['full name'].str.split(' ', expand = True)
# this is how full name column will be split into two seperate columns
```

```python
# Adding new columns and assigning value at the same time.
df[['first', 'last']] = df['full name'].str.split(' ', expand=True)
df
```

## Adding Row

```python
# New row as DataFrame, and other than 'fist' all values will be set to NaN.
new_row = pd.DataFrame({'first': ['Tony']})
df = pd.concat([df, new_row], ignore_index=True)
```

```python
df
```

## Removing Rows
Drop rows by index:

```python
df.drop(10, axis = 0, inplace=True)
# Removes the row with the specified index (use axis=0 for rows).
```

```python
df
```

# Drop multiple rows by index:
`df.drop([0, 2], axis=0, inplace=True)`

## Adding two DataFrames
### 1. Combine by Adding Rows (pd.concat)
Use case: You want one DataFrame under the other, with the same columns.

```python
people = {
  'first': ['Tony', 'Steve'],
  'last' : ['Stark', 'Rogers'],
  'email_address': ['tonystark@avenge.com', 'steverogers@avenge.com']
}

df2 = pd.DataFrame(people)
df2
```

```python
df = pd.concat([df, df2], ignore_index=True)
df
```
