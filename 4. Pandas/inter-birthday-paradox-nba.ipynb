# Birthday Problem Probability
The focus of this quick project is to put your Data Analysis with Pandas Skills to a test.

We're going to be analyzing the "Birthday Paradox" using data from the NBA.

In particular, the Birthday Paradox refers to the counterintuitive fact that only 23 people are needed for the probability of two people having a birthday on the same day, exceeds 50%.

This project will deal, not only with Pandas skills, but also with Probability concepts and combinatorics.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data/nba.csv')
pd.set_option('display.max_columns', 29)
```

```python
# Adding a new column Birth Date and adding random dates.

from datetime import datetime, timedelta

start_date = datetime.strptime('1985-01-01', '%Y-%m-%d')
end_date = datetime.strptime('2000-12-31', '%Y-%m-%d')

def random_dates(n, start, end):
    delta = end - start
    return [start + timedelta(days=np.random.randint(0, delta.days)) for _ in range(n)]

df['Birth Date'] = random_dates(len(df), start_date, end_date)

# Save to a new CSV
df.to_csv('data/nba.csv', index=False)
```

```python
# Deleting all the other columns and keeping only mentioned below.

df = df[['Player', 'Age', 'Pos', 'Birth Date']]

# Save back to the same CSV, overwriting it
df.to_csv('data/nba.csv', index=False)
```

```python
team_df = pd.read_csv('data/nba_team.csv')
team_df.shape
```

```python
# Adding Team column from the nba_team.csv to nba.csv

team_df = team_df[['Team']]
if len(df) != len(team_df):
    print("Warning: Number of players and number of teams do not match!")
df['Team'] = team_df['Team']
df.to_csv('data/nba.csv', index=False)
```

```python
# Deleting "*" from the end of the team name

df['Team'] = df['Team'].str.rstrip('*')
df.to_csv('data/nba.csv', index=False)
```
