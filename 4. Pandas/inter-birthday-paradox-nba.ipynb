## Birthday Probelm Probability
#### [Birthday Probability Problem](https://en.wikipedia.org/wiki/Birthday_problem)

##### The first step will be to calculate the probability of two people sharing a birthday, in a group of n people. We can use the following approximation formula:

![Birthday Probability Formuale](birthday-probability.png "Birthday Probability Formuale")

````python
def nCr(n ,k):
  f = math.factorial
  return f(n) / (f(k) * f( n - k))
````

````python
nCr(5, 2)
````

````python
nCr(6, 2)
````

#### 3. Write the function birthday_probability

Now it's time to generalize the probability calculation and create the function `birthday_probability` that returns the probability of two people sharing a birthday when in a group of `number_of_people`.

````python
def birthday_probability(number_of_people):
  return 1 - (364 / 365) ** nCr(number_of_people, 2)
````

# Interlude: Combinatorics

In this project, we use **combinatorics** to answer the question: _how many players share a birthday in a given team?_  

Combinatorics is a branch of mathematics that deals with **counting and arrangements**. Specifically, we are interested in **combinations**, which allow us to generate **all unique pairs** of items without caring about the order.

---

## Concept

Suppose you have `n` people and you want to check all **pairs** to see if they share the same birthday.  

- **Combinations formula:**  

\[
C(n, 2) = {n!}/{2!(n-2)!}
\]

- This generates all unique pairs `(Person A, Person B)` where order does not matter.
- Perfect for the **Birthday Paradox problem**, where we compare birthdays among all pairs.

---

## Example

Suppose we have three people:

| Name  | Birthday   |
|-------|------------|
| John  | March 5th  |
| Mary  | Sept 20th  |
| Rob   | March 5th  |

We want to find all pairs and see which share birthdays.

### Step 1: Generate all pairs of names

```python
from itertools import combinations

names = ["John", "Mary", "Rob"]
list(combinations(names, 2))
```

#### Output
```
[('John', 'Mary'), ('John', 'Rob'), ('Mary', 'Rob')]
```

### Step 2: Generate all pairs of birthdays

```python
birthdays = ["March 5th", "Sept 20th", "March 5th"]
list(combinations(birthdays, 2))
```

#### Output
```
[('March 5th', 'Sept 20th'), ('March 5th', 'March 5th'), ('Sept 20th', 'March 5th')]
```
✅ The second pair matches, meaning John and Rob share a birthday.

### Step 3: Put it in a DataFrame

```python
import pandas as pd

names_df = pd.DataFrame(combinations(names, 2), columns=["Person 1", "Person 2"])
birthdays_df = pd.DataFrame(combinations(birthdays, 2), columns=["Birthday 1", "Birthday 2"])

df_concat = pd.concat([names_df, birthdays_df], axis=1)
df_concat
```

#### Resulting DataFrame:
| Person 1 | Person 2 | Birthday 1 | Birthday 2 |
| :--------: | :--------: | :----------: | :----------: |
| John     | Mary     | March 5th  | Sept 20th  |
| John     | Rob      | March 5th  | March 5th  |
| Mary     | Rob      | Sept 20th  | March 5th  |

### Step 4: Check for shared birthdays

```python
df_concat['Birthday 1'] == df_concat['Birthday 2']
```

#### Output

```
0    False
1     True
2    False
```

✅ Only John and Rob share the same birthday.

#### 5. How many pairs of players share a birthday for the Atlanta Hawks?

This first activity will probably be the tougher one, as you have to find a way to answer how many players share a birthday for a given team. This first activity is related to the Atlanta Hawks. Hopefully, once you solve this one, you'll be able to answer the rest of the activities pretty easily.

>
> **Note:** We're asking for "pairs" of players, not the total number of players that share birthdays.  
> Imagine there are four players: A, B, C, and D.  
> - A and B have the same birthday → 1 pair  
> - C and D have the same birthday → 1 pair  
> So, there are **two pairs of players** with the same birthday: the first pair is A and B, and the second pair is C and D.
>

````python
atlanta_df = df.loc[df['Team'] == 'Atlanta Hawks']
atlanta_df.head()
````

````python
atlanta_df.shape
````

````python
name_df = pd.DataFrame(combinations(atlanta_df['Player'], 2), columns=['Player 1', 'Player 2'])
name_df
````

````python
birthday_df = pd.DataFrame(combinations(atlanta_df['Birthday'], 2), columns=['Birthday 1', 'Birthday 2'])
birthday_df
````

````python
check_df = pd.concat([name_df, birthday_df], axis=1)
check_df
````

````python
check_df['Birthday 1'] == check_df['Birthday 2']().sum()
````

````python
check_df['Birthday 1'] == check_df['Birthday 2']
````

````python
check_df.loc[(check_df['Birthday 1'] == check_df['Birthday 2'])]
````
