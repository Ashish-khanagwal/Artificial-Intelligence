# PANDAS

### What is pandas and how it differs from numpy and matplotlib
- Pandas is a Python library primarily used for data manipulation and analysis. It provides data structures and functions needed to handle structured data seamlessly.
- Numpy is mainly for numerical computing with multi-dimensional arrays and matrix operations.
- Matplotlib is a plotting library used for data visualization.
- While numpy handles numerical data arrays, pandas builds on it to provide richer data structures (DataFrame and Series) that handle labeled and heterogeneous data, making data analysis more intuitive and powerful.

```python
# Importing Pandas

import pandas as pd
```

## A DataFrame in pandas is a way to hold and work with structured data, similar to a table in a spreadsheet or a database.

### Beginner-Friendly Explanation
- Think of a DataFrame as a grid with rows and columns, like an Excel sheet.
- Each column in a DataFrame holds data of a specific type, such as numbers or text, and has a name like "Name" or "Age".
- Each row represents one record or entry, for example, one person's survey response.
- You can access, change, and analyze the data in this grid using simple commands without needing complex code.

### Key Points:
- DataFrames help organize data so it's easy to look up, sort, and manipulate.
- You can read data from files (like CSVs), preview it, filter it, and clean it using DataFrame methods.
- Pandas DataFrame makes working with large and messy data straightforward for beginners and professionals alike.

- Accessing columns and rows in a pandas DataFrame is intuitive, similar to working with a table in Excel. Here's a beginner-friendly explanation of each method:

#### Getting a Single Column as a Series
  - Use df['column_name'] to select just one column from your DataFrame.
  - The result is called a Series, which is like a single list of data from the table.



```python
ages = df.Age
ages

# This gives all values in the "Age" column.
```

#### Getting Multiple Columns
- Use `df[['col1', 'col2']]` with double brackets to select several columns at once.
- It returns another DataFrame with only those columns.

```python
subset = df[['Country', 'Age']]
subset
```

#### Getting Rows by Index
- Use `df.iloc[index]` to get a specific row by its position, starting from 0.
- For example, df.iloc gets the very first row.
- `df.iloc[0:5]` gets the first five rows as a DataFrame—great for previewing.

```python
first_row = df.iloc[0]
first_row
```

```python
first_five = df.iloc[0:5]
first_five
```

#### Filtering Rows by Condition
- To find rows that meet a specific requirement, write a condition in brackets.
- Example: `df[df['Age'] > 30]` returns only the rows where the "Age" value is greater than 30.

```python
older = df[df['Age'] > 30]
older
```
