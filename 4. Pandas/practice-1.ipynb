#### Why we did `.index`?
Our DataFrame has 'Word' as the index. When a column is the index, it is not available as a normal column, so doing 'df['Word']' will fail. And we get the error.

Instead, you should use the index itself.
- `words_most_common_value = df[df['Value'] == most_common_value].index.tolist()`

#### .tolist()
- Converts the resulting Series (the 'Word' column) into a Python list.
- So you get a simple list of words that have the most common value.

| Original line                                           | Why it fails                              | Correct approach                                      |
| :-------------------------------------------------------: | :-----------------------------------------: | :-----------------------------------------------------: |
| `df[df['Value'] == most_common_value]['Word'].tolist()` | `'Word'` is not a column (it’s the index) | `df[df['Value'] == most_common_value].index.tolist()` |

## 🔎 Why mode() caused an error and how to fix it
### My approach (why it failed)
```
most_common_ratio = df['Value Ratio'].mode()
words_common_ratio = df[df['Value Ratio'] == most_common_ratio].index.tolist()
```

- `df['Value Ratio'].mode()` returns a Series, not a single value.
 > 6.0    5

- When you do df['Value Ratio'] == most_common_ratio, you’re comparing:
  - `df['Value Ratio']` → a Series with as many rows as your DataFrame
  - `most_common_ratio` → a Series with just one row (index = 0)
- Pandas requires Series objects being compared to have the same index labels. Since they don’t match, you get:

`ValueError: Can only compare identically-labeled Series objects`

### Correct approach (why it works)
```
most_common_ratio = df['Value Ratio'].mode().iloc[0]
words_common_ratio = df[df['Value Ratio'] == most_common_ratio].index.tolist()
```
- `.iloc[0]` extracts the first element from the mode Series, giving you a scalar value (e.g., 10.0) instead of a Series.
- Now you are comparing:
  - `df['Value Ratio']` → Series of ratios
  - `most_common_ratio` → single scalar value
- This comparison works fine, and Pandas returns a Boolean mask where each row is checked against the scalar.
- The filter then selects all rows with that ratio, and `.index.tolist()` gives you the words.


| Approach                      | What it returns                       | Why it fails/works                                 |
| :-----------------------------: | :-------------------------------------: | :--------------------------------------------------: |
| `mode()` (your code)          | A **Series** (even if only one value) | Can’t compare Series with different labels → error |
| `mode().iloc[0]` (fixed code) | A **scalar** (single number)          | Scalar comparison works with Series → valid mask   |


````python
most_common_ratio = df['Value Ratio'].mode().iloc[0]
words_common_ratio = df[df['Value Ratio'] == most_common_ratio].index.tolist()
words_common_ratio
````

````python
df.query('`Value Ratio` == 6')
````

### 🔎 What it does:
- `.query()` lets you filter a DataFrame using a string expression instead of the usual Boolean mask.
- Inside the string, you can write conditions almost like plain Python.

So this line means:
👉 “Give me all rows where the column Value Ratio is equal to 6.”

##### Since your column name has a space ("Value Ratio"), Pandas normally expects you to use backticks around it:
```df.query('`Value Ratio` == 6')```

Without backticks, Pandas may get confused and throw an error like:

`UndefinedVariableError: name 'Value' is not defined`

```
# works if no spaces in column name
df.query('Value == 290')

# correct way when column has spaces
df.query('`Value Ratio` == 6')

# multiple conditions
df.query('`Value Ratio` == 6 and Char Count > 10')
```
