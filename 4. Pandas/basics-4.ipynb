```python
import pandas as pd
import numpy as np
```

```python
# Creating DataFrames using a dictionary
people = {
    "first": [
        "Ashish", "Corey", "John", "Meera", "Arjun",
        "Sophia", "Liam", "Hannah", "Carlos", "Jisoo", np.nan, None, 'NA'
    ],
    "last": [
        "Khanagwal", "Anderson", "Doe", "Patel", "Singh",
        "Williams", np.nan, "Nguyen", "Garcia", "Kim", np.nan, np.nan, 'Missing'
    ],
    "email_address": [
        "ashishkhanagwal@gmail.com",
        "coreyschafar@gmail.com",
        "timdavid@gmail.com",
        None,
        "arjunsingh@gmail.com",
        "sophiawilliams@gmail.com",
        "liambrown@gmail.com",
        "hannahnguyen@gmail.com",
        "carlosgarcia@gmail.com",
        "jisookim@gmail.com",
        None,
        np.nan,
        'Anonymous@email.com'
    ],
    "age": [24, 27, 26, 25, 29, 31, 28, 24, 32, 26, None, None, 'Missing'],
}
```

## Cleaning data: casting datatypes and handling missing values
Cleaning data in pandas typically starts with getting the right dtypes for each column and then dealing with missing values in a principled way. Correct dtypes make operations faster and safer, while robust missing-value handling prevents errors and bias in analysis.

```python
df = pd.DataFrame(people)

# To handle 'Missing' or any 'arbitrary' value we can do use numpy to replace those with np.nan.
df.replace('NA', np.nan, inplace=True)
df.replace('Missing', np.nan, inplace=True)
df
```

#### These show memory use, types, and where missing values appear.

```python
df.dtypes
```

```python
df.isna().sum()
```

```python
# df.dropna()
# drop any row with at least one NaN
# df.dropna(axis='index', how='any) -> This will drop the rows with at least one NaN.
# If we change 'how='any'' to 'all' then it will drop the row with all NaN values
df.dropna(axis='index', how='all')
```

```python
df.dropna(axis='index', how='all', subset=['email_address'])
# this will drop the rows with NaN value in the series 'email_address'
```

```python
df.dropna(axis='index', how='all', subset=['last', 'email_address'])
# This command removes rows where both last and email_address are missing, but keeps rows where at least one of them has a value.
```

```python
df.isna()

# It checks every cell in your DataFrame.
# If the value is missing (NaN), it returns True.
# If the value is present, it returns False.
# So the output is a Boolean DataFrame of the same shape as your original one.
```

```python
df.fillna(0)
# It will fill all the NaN values with 0.
```
