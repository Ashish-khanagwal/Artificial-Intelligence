## A Series in pandas is best understood as a single column of labeled data, like one list or array in a spreadsheet.

### Beginner-Friendly Explanation
- Imagine a Series as a vertical list of data values, each with a label called the index.
- Each value in the Series can be a number, string, or other type, and every value has a unique position or index on the left side, like row numbers in Excel.
- For example, if you have survey answers or ages, those could be stored in a Series.

### Key Points:
- A Series is simpler than a DataFrame; it contains data in just one column but with many rows.
- You can access items using their position (like series) or their label (like series['Alice']).
- Series makes it easy to analyze, process, and manipulate one set of data at a time.

Series is the foundation for columns in a DataFrame. Whenever you select a single column from a DataFrame, you get a Series. This makes Series crucial for all kinds of fast, efficient, and beginner-friendly data operations.

#### The main difference between df.loc and df.iloc in pandas is how you select rows and columns—by label or by position.

### `df.loc`: Label-Based Indexing
- `df.loc` is used to access data by row and column labels (names).
- If your DataFrame has custom indexes like names or specific labels, you use those with `loc`.
- Slicing with `.loc` is inclusive of the end label.

Example:
```
df.loc['B', 'Age']      # Selects the row with label 'B' and column 'Age'
df.loc['A':'C']         # Selects rows from label 'A' to 'C' (inclusive)
```

### `df.iloc`: Integer-Based Indexing
- `df.iloc` is used to access data by row and column positions, using integer indices (like you would with a Python list).
- The first row and column are at position 0, the second at 1, and so on.
- Slicing with `.iloc` is exclusive of the end index, just like standard Python lists.

Example:
```
df.iloc[1, 1]           # Selects the second row and second column (both index 1)
df.iloc[0:2]            # Selects rows at positions 0 and 1 (end excluded)
```

##### Indexes in pandas are labels for rows that help organize, access, and manage DataFrames efficiently. Here’s a beginner-friendly, in-depth guide with practical examples:

### Understanding Indexes
- By default, every DataFrame row gets an index starting from 0, 1, 2, etc. This is helpful for referencing rows.
- Indexes can be changed to use meaningful labels, like names or dates, making data easier to work with.

### Setting an Index
- Use `df.set_index('ColumnName')` to make a column the index. The original row numbers are replaced by those values.

- You can also set multiple columns as a "MultiIndex" using a list: `df.set_index(['Name', 'Age'])`

````python
df.set_index('last', inplace = True)
df
````

### Resetting an Index
- To restore the default numeric index, use df.reset_index(), which puts the current index back as a column and re-numbers rows from zero.

```
df_reset = df.reset_index()
print(df_reset)
```

- If you want to discard the old index, use `df.reset_index(drop=True)`.
- You can do this in place with inplace=True: `df.reset_index(inplace=True)`

````python
df.reset_index(inplace = True)
df
````

````python
df.set_index('first', inplace= True)
````

#### Single Cell: Use .at (by label) or .iat (by position) to update one specific value.

````python
df.at[1, 'last'] = 'Anderson'
df
````

## 1. What .loc does
- .loc is like a Swiss army knife — it can select:
  - single cells
  - entire rows
  - entire columns
  - a block of rows and columns

#### So when you do something like:

```
df.loc[2, 'Score'] = 95
```

- it will work — pandas will update that one cell.
- But behind the scenes, pandas first checks: "Am I selecting one value, or a slice of the DataFrame?"
That checking makes `.loc` a bit slower for single-cell updates.

## 2. What .at does
- .at is specialized: it is made only for getting or setting a single cell (scalar).
```
df.at[2, 'Score'] = 95
```

- Here, pandas knows right away:

👉 "Oh, this is just one specific cell. No need to check for slices or ranges."

#### That makes `.at`:
- faster
- clearer in intention (other programmers immediately know you’re working with one single value, not a row or slice)

````python
df
````

````python
df.iat[2, 3] = 28
# 2nd row, and 3rd column -> Age of John
df
````

````python
df.iloc[2, [0, 1, 3, 5]] = ["Johny", "Bairstow", 30, 'Amazon']
df
````

### 🔑 Quick Summary:
- .at → label-based, fast for single cell.
- .iat → position-based, fast for single cell.
- .loc → label-based, flexible (rows, slices, cells, multiple).
- .iloc → position-based, flexible (rows, slices, cells, multiple).

## Difference between label based and position based:

#### `df.iloc[1, ['last', 'age', 'Company']]` -> Says (TypeError: Cannot index by location index with a non-integer key) 
## Why above `.iloc` code failed?
- `.iloc` is position-based only (integer row & column positions).
- But here, you gave it `['last', 'age', 'Company']`, which are column labels (strings).
- That’s why pandas complains:
“Cannot index by location index with a non-integer key”

#### `df.loc[1, ['last', 'age', 'Company']]` -> This will work as expected.
## Why `.loc` works
- `.loc` is label-based.
- You’re using 1 (row label) and `['last', 'age', 'Company']` (column labels).
- That matches perfectly → so pandas updates the row as expected.

### 1. apply: Row/Column-wise Operations
- Purpose: Used for applying a function to each row or column of a DataFrame, or to each value in a Series.
- How it works: It processes one entire row or column at a time.

````python
df['email'].apply(len)
````

````python
def upper_email(email):
  return email.upper()

df['email'] = df['email'].apply(upper_email)
df
````

````python
df['email'] = df['email'].apply(lambda x: x.lower())
df
````

````python
df['Status'] = df['score'].apply(lambda x: 'Pass' if x > 90 else 'Fail')
df
# Update with a Formula or Function: Apply math operations or functions to update all values.
````

````python
# This is how apply works on dataframe
df.apply(len)
# Actually, this is telling is the length each label (how many values stored).
````

````python
df.apply(len, axis=1)
# And this is telling us the number of elements stored in each row.
````

````python
df.apply(pd.Series.min)
# This will give the smallest values in the whole dataframe from all series.
````

#### `DataFrame.apply(func, axis=0)` → applies func column by column (default axis=0).
#### `DataFrame.apply(func, axis=1)` → applies func row by row.

````python
df.apply(lambda x: x.max())
# Because of above logic, that's why here 'x' refer to series.
````

#### running apply to a series, applies a functions to every value in the series.
#### running apply to a dataframe, applies the function to every series in the dataframe.

````python
df[['age', 'score']] = df[['age', 'score']].apply(pd.to_numeric)
````
