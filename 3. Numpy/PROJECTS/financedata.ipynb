# Sample Finance Data (Annual Profit/Loss)
- Columns: [company_id, 2020, 2021, 2022, 2023, 2024]
- Rows: Each row is one company’s yearly profit/loss (can include negative values to indicate losses).

```python
import numpy as np
```

```python
finance_data = np.array([
    [1, 50000, 75000, -12000, 20000, 65000],   # Company A
    [2, 120000, -60000, 95000, 102000, 90000], # Company B
    [3, -20000, 40000, 45000, 70000, 75000],   # Company C
    [4, 260000, 310000, 290000, 280000, 265000],  # Company D
    [5, 150000, 152000, 155000, 155000, 160000],  # Company E
    [6, 80000, 120000, 115000, 117000, 119000]    # Company F
])
```

```python
company_id = finance_data[:, 0]
values = finance_data[:, 1:]
```

```python
# 1. Total profit/lose for each company across years
total_profit = np.sum(values, axis = 1)
print("Total profit/loss per company:\n", total_profit)
```

```python
# 2. Year-over-year changes (differences) and growth rates

yearly_change = np.diff(values, axis = 1)
growth_rate = yearly_change / values[:, :-1] * 100

print("Year over year changes:\n", yearly_change, "\n")
print("Year over year growth rate (%)\n", growth_rate)
```

```python
# 3. Identify companis with losses (negative numbers) any year
loss_mask = np.any(values < 0, axis = 1)
companies_losses = company_id[loss_mask]

print("Company with losses in any year:\n", companies_losses)
```

```python
# 4. Rank companies by overall performance (total profit descending)
rank_indices = np.argsort(total_profit)[::-1]
rank_companies = company_id[rank_indices]
rank_total = total_profit[rank_indices]

print("Ranked companies by total profit:\n", rank_companies)
print("Corresponding total profit:\n", rank_total)
```

```python
# 5. Detect volatility (std deviation of profits across years)
volatility = np.std(values, axis = 1)
print("Volatility (standard deviation) per company:\n", volatility)
```

```python
# 6. Mask/filter companies with losses in two or more years
loss_counts = np.sum(values < 0, axis = 1)
mask_two_or_more_losses = loss_counts >= 2
companies_with_multiple_losses = company_id[mask_two_or_more_losses]
print("Companies with two or more losses:\n", companies_with_multiple_losses)
```

```python
# 7. Forecast next year (2025) profit using simple linear extrapolation
# Using difference between last two years to estimate next year
last_year_diff = values[:, -1] - values[:, -2]
forecast_2025 = values[:, -1] + last_year_diff
print("Forecasted 2025 profit:\n", forecast_2025)
```

```python
# 8. Visualize trends (profit turning to loss or vice versa)
# Find year pairs where sign changes between consecutive years
sign_changes = np.diff(np.sign(values), axis = 1) != 0
companies_sign_change = company_id[np.any(sign_changes, axis = 1)]
print("Companies with profit/loss sign change:\n", companies_sign_change)
```
