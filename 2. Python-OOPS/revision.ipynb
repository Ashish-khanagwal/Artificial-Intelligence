```python
class dog:
  legs = '4 legs'

  def __init__(self, name, age):
    self.name = name
    self.age = age

  def bark(self):
    print(f"{self.name} of age {self.age} woof!")

dog1 = dog('Muku', 5)
dog1.bark()
print(dog1.legs)
```

```python
# Bank System

class bank_system:
  def __init__(self, name, balance=0):
    self.name = name
    self.balance = balance

  def deposit(self, amount):
    self.balance += amount
    print(f"{amount} is deposited in your account, New balance is {self.balance}")

  def withdraw(self, amount):
    if amount > self.balance:
      print("Insufuicient Funds!")
    else:
      self.balance -= amount
      print(f"{amount} is withdrawned from account, New balance is {self.balance}")

  def get_amount(self):
    return self.balance
  
customer1 = bank_system('Ashish', 1000000)
customer1.balance
```

```python
customer1.deposit(50000)
```

```python
customer1.withdraw(40000)
```

```python
customer1.balance
```

```python
customer1.get_amount()
```

```python
# Inheritance (Single)

class Car:
  def __init__(self, window, door, enginetype):
    self.window = window
    self.door = door
    self.enginetype = enginetype

  def drive(self):
    print(f"The person will drive the {self.enginetype} car.")

class Tesla(Car):
  def __init__(self, window, door, enginetype, is_electric):
    super().__init__(window, door, enginetype)
    self.is_electric = is_electric

  def self_driving(self):
    print("Is this car is self driving: {self.is_electric}")

tesla1 = Tesla(4, 2, 'electric', True)
print(tesla1.window)
print(tesla1.door)
print(tesla1.enginetype)
tesla1.self_driving()
```

```python
## Multiple Inheritance

class Animal:
  def __init__(self, name):
    self.name = name

  def speak(self):
    print("Subclass must implement this method")

class Pet:
  def __init__(self, owner):
    self.owner = owner

class Dog(Animal, Pet):
  def __init__(self, name, owner):
    Animal.__init__(self, name)
    Pet.__init__(self, owner)

  def barks(self):
    return f"{self.name} says woof!"
  
dog1 = Dog('muku', 'sam')
print(dog1.name)
print(dog1.owner)
print(dog1.barks())
```

```python
# Polymorphism

# Over riding
class Animal:
  def speak(self):
    return "Speaks"
  
class Dog(Animal):
  def speak(self):
    return 'Woof'
  
class Cat(Animal):
  def speak(self):
    return 'Meow'
  
def animal_speak(animal):
  print(animal.speak())

dog = Dog()
cat = Cat()
print(dog.speak())
print(cat.speak())
animal_speak(dog)
```
