# ⚙️ PART 1: Advanced Customization in Matplotlib
### 🎯 1. Figure Size & Resolution

#### By default, Matplotlib creates small 6x4 inch figures. You can change this easily.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([2, 4, 8, 32])

plt.figure(figsize=(8, 5), dpi=240)
plt.plot(x , y)
plt.title("Custom Figure Size & Resolution")
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

### 🧭 2. Adding Gridlines
#### Gridlines make your chart easier to read — especially for numeric data.

```python
plt.plot(x, y)
plt.title('with gridlines')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

##### 👉 alpha controls transparency.
##### 👉 You can also use `plt.minorticks_on()` for smaller gridlines.

## ✏️ 3. Annotating Specific Points

#### Annotations are useful for marking key data points (like peaks, or anomalies).

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 7, 1, 0]

plt.plot(x, y, marker='o')
plt.title("Annotations Example")

plt.annotate("Peak Point", xy=(3, 7), xytext=(3.5, 6),
            arrowprops=dict(facecolor='green', arrowstyle='->', linewidth=2))
plt.show()

# 🪶 The arrowprops adds an arrow, making your chart more informative.
```

## 🤝 PART 2: Matplotlib + Pandas Integration

#### When you’re working with real-world datasets, it’s easier to plot directly from a Pandas DataFrame — no need to manually extract arrays.

```python
data = {
    'Year': [2019, 2020, 2021, 2022, 2023],
    'Sales': [120, 150, 170, 200, 250]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.plot(x='Year', y='Sales', kind='line', marker='o', color='green', title='Company Sales Over Time')
plt.show()
```

## 📊 2. Bar Chart from DataFrame

```python
df.plot(x='Year', y='Sales', kind='bar', color='orange', title='Sales Bar Chart')
plt.show()
```

## 3. Multiple Columns

```python
data = {
    'Year': [2019, 2020, 2021, 2022, 2023],
    'Revenue': [120, 150, 170, 200, 250],
    'Profit': [20, 25, 30, 40, 50]
}
df = pd.DataFrame(data)

df.plot(x='Year', y=['Revenue', 'Profit'], kind='line', marker='o')
plt.title("Revenue vs Profit")
plt.show()
```

## 📈 4. Histogram and Boxplot from DataFrame

```python
df['Revenue'].plot(kind='hist', bins=10, color='skyblue', edgecolor='black', title='Revenue Distribution')
plt.show()
```

```python
df[['Revenue', 'Profit']].plot(kind='box', title='Revenue vs Profit Spread')
plt.show()
```
