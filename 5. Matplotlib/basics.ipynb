# 🌈 Basics of Matplotlib
## 🔹 What is Matplotlib?

### - Matplotlib is the most popular plotting library in Python, used to visualize data as graphs, charts, and figures.

### - Think of it as your way to see your data — trends, distributions, comparisons — instead of just reading numbers.

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
plt.plot([1, 2, 3], [7, 6, 5])
plt.show()
```

## 🪄 Step 1: Your First Plot

Here’s the absolute simplest example 👇

```python
x = [2023, 2024, 2025, 2026]
y = [15, 25, 30, 35]

plt.plot(x, y)
plt.show()
```

## 🔍 What’s Happening:

- `plt.plot()` — draws a line graph.
- `plt.show()` — renders the plot window.
- The x list is your X-axis values.
- The y list is your Y-axis values.

✅ You just made your first graph 🎉

## 🪶 Step 2: Adding Titles and Labels

You can make your chart more meaningful using titles and axis labels:

```python
plt.plot(x, y)
plt.title('Simple line plot')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.show()
```

## 🎨 Step 3: Styling the Line

You can change color, style, and markers easily:

```python
plt.plot(x, y, color='red', linestyle='--', marker='o')
plt.title("Styled Line Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

| Property    | Example                      | Meaning     |
| :-----------: | :----------------------------: | :-----------: |
| `color`     | `'r'`, `'blue'`, `'#00FF00'` | Line color  |
| `linestyle` | `'-'`, `'--'`, `':'`         | Line type   |
| `marker`    | `'o'`, `'^'`, `'s'`          | Dots/points |

## 🧱 Step 4: Multiple Lines in One Plot

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [2, 3, 4, 5, 6]

plt.plot(x, y1, label="Squares")
plt.plot(x, y2, label="Linear")
plt.title("Multiple Lines Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()  # shows the label box
plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

plt.plot(x, y)
plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3= np.array([13, 15, 20, 30])

line_style = dict(marker='.',
          markersize=20,
          markerfacecolor='#ff3686',
          markeredgecolor='#59002b',
          linestyle='solid',
          linewidth=5,
          )

plt.plot(x, y1, color='#9c1154', **line_style)

plt.plot(x, y2, color='#42c773', **line_style)

plt.plot(x, y3, color='#4270c7', **line_style)

plt.show()
```

## Bar Charts

#### Compare categories of data by representing each category with a bar.

```python
categories = ['Grains', 'Fruits', 'Vegetables', 'Protein', 'Dairy', 'Sweets']
values = np.array([10, 24, 36, 29, 16, 40])

plt.bar(categories, values)

plt.title('Daily Consumption')
plt.xlabel('Food')
plt.ylabel('Quantity')

plt.show()
```

## Scatter Plot

#### Great for showing relationships between two variables.
#### Helps to identify a correlation (+, -, None)
#### Example: Study hours vs Test scores

```python
x1 = np.array([0, 1, 1, 2, 3, 4, 5, 6, 7, 7, 8])
y1 = np.array([50, 60, 65, 62, 68, 70, 75, 78, 82, 85, 87])

x2 = np.array([0, 1, 2, 2, 3, 4, 5, 6, 7, 8, 8])
y2 = np.array([45, 56, 60, 64, 69, 73, 78, 82, 85, 90, 95])

plt.scatter(x1, y1, color='r',
            alpha=0.5,
            s = 50,
            label='Class A')

plt.scatter(x2, y2, color='green',
            alpha=0.5,
            s = 50,
            label='Class B')

plt.title('Test Scores')
plt.xlabel('Hours Studied')
plt.ylabel('Grade')

plt.legend()
plt.show()
```

## Pie Chart
#### Circular chart divided into slices to show the percentages of the total.
#### Good for visualizing distribution among categories.

```python
categories= ['Freshers', 'Sophomores', 'Juniors', 'Seniors']
values = np.array([300, 250, 275, 225])
colors=['red', 'yellow', 'blue', 'green']

plt.pie(values, 
      labels=categories,
      autopct='%1.1f%%',
      colors=colors,
      explode = [0, 0, 0, 0.1],
      shadow=True,
      startangle=90)

plt.title('Bro Code College')

plt.show()
```

## Histograms

#### A visual representation of the distribution of quantitative data.
#### They group values into bins (intervals) and counts how many fall in each range.

```python
scores = np.random.normal(loc=80, scale=10, size=100)
scores = np.clip(scores, 0, 100)

plt.hist(scores, bins=10,
        color='lightgreen',
        edgecolor='black')

plt.title('Exam Scores')
plt.xlabel('Score')
plt.ylabel('No. of Students')

plt.show()
```

```python
data = np.random.randn(1000)  # random 1000 numbers
plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title("Histogram Example")
plt.show()
```

## Subplots
#### Figure = The entire canvas
#### Ax = A single plot (subplot)

```python
x = np.array([1, 2, 3, 4, 5])

figure, axes = plt.subplots(2, 2) # A canvas of 2 rows and 2 columns (plots)

axes[0,0].plot(x, x*2, color='red')
axes[0,0].set_title('x*2')

axes[0,1].plot(x, x**2, color='blue')
axes[0,1].set_title('x**2')

axes[1,0].plot(x, x**3, color='green')
axes[1,0].set_title('x**3')

axes[1,1].plot(x, x**4, color='purple')
axes[1,1].set_title('x**4')

plt.tight_layout()
plt.show()
```
